Company merger : In corporate acquisitions, buyers either execute a 100% take-private buyout (delisting the target company by buying all shares via tender offers and forcing out remaining holdouts through a "squeeze-out" merger) or acquire a majority controlling stake (leaving the target listed as a publicly traded subsidiary but with reduced share supply). Both methods require the buyer to pay an acquisition premium above the current market value, which triggers an instant stock price spike that traps short sellers into forced, massive cash-settled losses and automatically converts options contracts into cash or new stock. Once the deal closes, the acquiring company wields total voting control, allowing them to completely replace the board of directors and executive leadership to dictate the company's new strategic direction.
